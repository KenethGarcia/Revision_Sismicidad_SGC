This notebook will make some tests to check if the package is working correctly. We will start by doing some simple TOML files to check if the package is able to read them correctly. Then we will move on to more complex tests.

To initialize any object in the package, you will need to create a TOML file with the configuration of the object you want to create. The package will read this file and create the object accordingly.

````python
from src.core import runner as pyrsnc

runner_obj = pyrsnc.Runner(config_path="test.toml")
````

At the initialization of the Runner object, the package will read the TOML file and create the object accordingly. If the TOML file is not found or is not valid, the package will raise an error. Let's start with a simple TOML file to check if the package is able to read it correctly. The TOML file will contain the configuration of a simple object that we will create.

Any TOML file must have at least the following info:
- A database provided
- A query to be executed on the database

```toml
[[database]]
name       = "sc6"
env_file   = ".env"  # It requires the .env file at the same path as the TOML file, otherwise specify the full path to the .env file
env_prefix = "SERVER_SC6"

[[queries]]
name       = "sc6"
sql_file   = "./queries/revision.sql"
description= "Query to retrieve events from SeisComP6 database"
database   = "sc6"
```
