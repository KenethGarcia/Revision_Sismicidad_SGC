## 1. Making queries to the database

Probably one of the most important aspects in the sismicity revision is the ability to make queries to the database. This is a very important aspect because it allows us to extract the data that we need to perform the analysis. In the revision routine, it has been used the `pymysql` library to make queries to the database. You can read more about this library [here](https://pymysql.readthedocs.io/en/latest/).

Let's start by importing the library:

```python
import pymysql
```

`pymsql` requires the following parameters to connect to the database:
- host: the host of the database
- user: the user to connect to the database
- password: the password of the user
- database: the name of the database
- port: the port of the database

In this case, we have the following parameters:

```python
connection = pymysql.connect(host="172.25.3.135", # This is the IP address of the database
                             user="consulta", # This is the user to connect to the database
                             passwd="consulta", # This is the password of the user
                             db="seiscomp3") # This is the name of the database
```


Now, with the connection established, we can make queries to the database. Here we will use pandas to store the results of the queries in a DataFrame. You can read more about pandas [here](https://pandas.pydata.org/). In particular, the method `read_sql` allows us to read the results of a query directly into a DataFrame. This function requires as argument the query and the connection to the database. The queries can be a little bit complex, but in general, they are simple. For example, the following query retrieves the first 10 rows of the table `event`:

```python
first_rows = pd.read_sql("SELECT * FROM Event LIMIT 10", connection)
first_rows
```

And the last 20 rows of the table `event`, containing the last 20 events processed by analysts or automatic systems:

```python
last_rows = pd.read_sql("SELECT * FROM Event ORDER BY _oid DESC LIMIT 20", connection)
last_rows
```

As you can notice, the table have the following columns:

```python
print([i for i in last_rows.columns])
```

But, how we can access to more global parameters, such as the uncertainty of the location of the events? This information is stored in the table `Origin`. The following query retrieves the last 10 rows of the table `Origin`:

```python
origin_example = pd.read_sql("SELECT * FROM Origin ORDER BY _oid DESC LIMIT 10", connection)
origin_example
```

This table contains all the required columns to create an origin in seiscomp3. For the purpose of the sismicity revision, we are interested in the following columns:
- `_oid`: the unique identifier of the origin
- `time_value`: the origin time of the event
- `latitude_value`: the latitude of the event
- `latitude_uncertainty`: the uncertainty of the latitude
- `longitude_value`: the longitude of the event
- `longitude_uncertainty`: the uncertainty of the longitude
- `depth_value`: the depth of the event
- `depth_uncertainty`: the uncertainty of the depth
- `methodID`: the method used to calculate the origin
- `earthModelID`: the earth model used to calculate the origin
- `quality_associatedPhaseCount`: the number of TOTAL associated phases
- `quality_usedPhaseCount`: the number of used phases for localization
- `quality_usedStationCount`: the number of used stations for localization
- `quality_azimuthalGap`: the azimuthal gap of the localization
- `creationInfo_author`: the author of the origin, i.e. the analyst who created the origin
- `creationInfo_creationTime`: the creation time of the origin

**WARNING**: The table `Origin` contains all the origins without any filter. It means that, for a unique event, there can be multiple origins. Notice that some _oid have the same time value.

Similar to the Event and Origin tables, there is a lot of information stored in the database. You can see all the tables contained in the database by running the following query:

```python
# Print all tables contained in the database
tables = pd.read_sql("SHOW TABLES", connection)
tables
```

For example, the table Magnitude contains the magnitude information of the events. The following query retrieves the last 10 rows of the table `Magnitude`:

```python
magnitudes = pd.read_sql("SELECT * FROM Magnitude ORDER BY _oid DESC LIMIT 10", connection)
magnitudes
```
